```python
#1. input section
print("Enter your weekly goal (fl oz):")
goal = int(input())
print("Enter daily water intake (fl oz):")
sunday = int(input("Sunday: "))
monday = int(input("Monday: "))
tuesday = int(input("Tuesday: "))
wednesday = int(input("Wednesday: "))
thursday = int(input("Thursday: "))
friday = int(input("Friday: "))
saturday = int(input("Saturday: "))

#2. data processing section
total_intake = sunday + monday + tuesday + wednesday + thursday + friday + saturday
average_intake = total_intake / 7
rounded_average_intake = round(average_intake)

#3. output section
if total_intake >= goal:
  print("You reached your goal!")
else:
  print("You failed to meet your goal, try harder next time.")

print("Total weekly intake:", total_intake, "fl oz")
print("Goal:", goal)
print("Average daily intake:", rounded_average_intake)
```
